# Leakage-free contract prediction

This notebook replaces the original client-scoring experiment. The previous version used `Стадия работы с клиентом` both to define the outcome and as a feature. That is post-outcome information and causes target leakage.

The corrected benchmark creates a synthetic lead-level outcome from variables available before the contract decision, then keeps the stage field only as a post-outcome reference.

```python
from pathlib import Path
import sys
sys.path.insert(0, str(Path.cwd().parent / 'src'))
from contract_prediction import make_synthetic_modeling_data, train_and_evaluate, out_of_fold_scores
```

```python
df = make_synthetic_modeling_data(n_clients=1000, seed=42)
df.head()
```

## Feature policy

Allowed features: company type, client status, company profit and product need. The current work stage is deliberately excluded because it is known only after the sales process has progressed.

```python
result = train_and_evaluate(df)
result.metrics
```

```python
scores = out_of_fold_scores(df)
scores.sort_values('Вероятность заключения договора', ascending=False).head(10)
```

## Interpretation

The hold-out metrics describe model performance on unseen records. The out-of-fold table provides a more defensible portfolio scoring snapshot for all synthetic clients. Neither should be interpreted as validation on real company data.
